```python
import numpy as np
import pandas as pd

# 1. Load the dataset directly from Google Drive link
url = 'https://drive.google.com/uc?export=download&id=1paSF8OTsp4K4tamGMx3BhaC_MbO4y5ob'
df = pd.read_csv(url)

print('Shape before cleaning:', df.shape)
print('\nMissing values before cleaning:')
print(df.isna().sum())

# 2. Remove exact duplicate rows (and duplicate Customer_IDs if any)
df = df.drop_duplicates().copy()
df = df.drop_duplicates(subset=['Customer_ID'], keep='first').copy()

# 3. Clean text columns: strip extra whitespace
for col in ['Customer_ID', 'Customer_Name', 'Region', 'Customer_Segment']:
  df[col] = df[col].astype(str).str.strip()

# 4. Standardize casing for Region and Customer_Segment (e.g., 'mansoura' -> 'Mansoura', 'CAIRO' -> 'Cairo')
df['Region'] = df['Region'].str.title()
df['Customer_Segment'] = df['Customer_Segment'].str.title()

# 5. Handle any potential missing values
df = df.replace(
    {'nan': 'Unknown', 'None': 'Unknown', '': 'Unknown'}
)

# 6. Final check and export
print('\nShape after cleaning:', df.shape)
print('\nUnique Regions after cleaning:', df['Region'].unique())
print('Unique Segments after cleaning:', df['Customer_Segment'].unique())
display(df.head(10))

df.to_csv('cleaned_customers_master.csv', index=False)
print('\nDone! Saved to cleaned_customers_master.csv')
```
