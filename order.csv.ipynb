```python
import io
import numpy as np
import pandas as pd

# Sample dataset inside the code so no external file is needed
csv_data = """id,name,age,salary,city
1, Alice ,25,50000, Cairo
2,Bob,,60000,Alexandria
2,Bob,,60000,Alexandria
3,Charlie,30,,
4, na ,28,55000,Giza
"""

df = pd.read_csv(io.StringIO(csv_data))

print('Shape:', df.shape)
display(df.head())
print('\nMissing values before cleaning:')
print(df.isna().sum())

df = df.drop_duplicates().copy()

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.lower()
    .str.replace(r'\s+', '_', regex=True)
)

obj_cols = df.select_dtypes(include=['object', 'string']).columns
for col in obj_cols:
  s = df[col].astype('string').str.strip()
  s = s.replace(
      [
          '',
          'nan',
          'NaN',
          'None',
          'none',
          'NULL',
          'null',
          'N/A',
          'n/a',
          'na',
          'NA',
      ],
      pd.NA,
  )
  df[col] = s.fillna('Unknown').astype('object')

num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
  if df[col].notna().any():
    df[col] = df[col].fillna(df[col].median())
  else:
    df[col] = df[col].fillna(0)

print('\nMissing values after cleaning:')
print(df.isna().sum())
print('\nCleaned data:')
display(df)
```
