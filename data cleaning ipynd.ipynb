```python
from google.colab import files
import pandas as pd

uploaded = files.upload()
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("Loaded file:", filename)
print("Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nMissing values:")
print(df.isna().sum())
```

```python
# Clean text columns
text_cols = ["Category", "Title", "In_Stock"]
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()

df["Category"] = df["Category"].str.title()
df["In_Stock"] = df["In_Stock"].str.title()

# Clean Price column and convert to numeric
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("£", "", regex=False)
    .str.strip()
)
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

# Remove duplicates
before_rows = len(df)
df = df.drop_duplicates().copy()
after_rows = len(df)

# Save cleaned dataset
cleaned_filename = "cleaned_books.csv"
df.to_csv(cleaned_filename, index=False, encoding="utf-8-sig")

print("Removed duplicates:", before_rows - after_rows)
print("Final shape:", df.shape)
print("\nData types:")
print(df.dtypes)
print("\nCleaned sample:")
print(df.head())
```
